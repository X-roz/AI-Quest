## Images with NUMPY 

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# loading image
pic = Image.open("dataset/img/zoro.jpg")
pic
```

```python
# creating a numpy arr with the picture loaded
pic_arr = np.asarray(pic)
pic_arr.shape
```

```python
plt.imshow(pic_arr)
plt.show()
```

```python
pic_rgbcolur = pic_arr.copy()

# the picture shape is (2488, 1399, 3) with 3 dimensional/channel
# 0 - Red
# 1 - Green
# 2 - blue
# Each channel has range of values to represent the colors. 

# Red channel values 0 - 255 range
# 0   --> no color (pure black)
# 255 --> full color (pure red)
plt.imshow(pic_rgbcolur[:,:, 0], cmap = 'gray')
plt.show()
```

```python
# Green channel
plt.imshow(pic_rgbcolur[:,:, 1], cmap='gray')
plt.show()
```

```python
# blue channel
plt.imshow(pic_rgbcolur[:,:, 2], cmap = 'gray')
plt.show()
```

```python
# we are going to show only red without any green or blue channels in the image.
# Take all rows and columns of the green channel and update its as zero.
# By this way we are completely supressing green channel.
# similar for blue channel
pic_rgbcolur[:,:, 1] = 0 
pic_rgbcolur[:,:, 2] = 0
plt.imshow(pic_rgbcolur)
plt.show()
```
